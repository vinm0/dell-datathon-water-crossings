# 4. Where would a water crossing help?

**Idea:** use before/after top-down pairs (as in notebook 3) and ask the VLM for *structured* output: flooded roads, isolated buildings and a crossing-need level. Rank locations, add rainfall context, then review the top candidates by eye.

Data: `datasets/Texas_Flood_Student_Data/` (see `texas_flood.py`). Model answers are suggestions to inspect, not annotations.

```python
from texas_flood import load_texas_images, show_images
from crossing_analysis import analyze_locations, add_rainfall

# (before, after) pairs; same IDs as notebook 2
ids = ['TX001','TX002','TX003','TX004','TX011','TX012','TX013','TX018','TX019','TX020']
pairs = [(f'{i}_2016_NAIP', f'{i}_2017_NOAA') for i in ids]
```

## Analyze each pair
One model call per pair. `raw` keeps the model's full reply so you can check the JSON was read correctly.

```python
table = analyze_locations(pairs)
table[['after_id','locality','crossing_need','flooded_roads','isolated_buildings','road_cut_off','uncertainty']]
```

## Add rainfall context
Total rain in the 14 days before the response image. Rain does not prove what is visible; it helps prioritize.

```python
table = add_rainfall(table, days_before=14)
table[['after_id','locality','crossing_need','rain_mm_total']]
```

## Review the top candidates
Look at the highest-need pairs and read the model's stated locations. Reject any the images do not support.

```python
top = table.head(3)
for _, row in top.iterrows():
    show_images(load_texas_images([row['before_id'], row['after_id']]))
    print(row['after_id'], row['crossing_need'], row['locations'])
    print('Uncertainty:', row['uncertainty'], '
')
```
